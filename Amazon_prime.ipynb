# **DATA CLEANING**

**1. Remove duplicates from Titles**

```python
titles = titles.drop_duplicates()
print("Duplicates:", titles.duplicated().sum())
```

```python
credits = credits.drop_duplicates()

print("Duplicates:", credits.duplicated().sum())
```

**3. Fill missing descriptions**

```python
titles['description'] = titles['description'].fillna('No Description')
```

**4. Fill missing Age Certification**

```python
titles['age_certification'] = titles['age_certification'].fillna('Not Rated')
```

```python
shows = titles[titles['type'] == 'SHOW']
```

**5. IMDb Score and IMDb Votes**

```python
rating_data = titles.dropna(subset=['imdb_score'])
```

**6. Clean missing Character in Credits**

```python
credits['character'] = credits['character'].fillna('Unknown')
```

# **Data Visualization**

**Movies vs TV Shows**

```python
type_count = titles['type'].value_counts()

fig = px.pie(
    values=type_count.values,
    names=type_count.index,
    title='Movies vs TV Shows'
)

fig.show()
```

**Content Released by Year**

```python
year_count = titles['release_year'].value_counts().sort_index()

fig = px.line(
    x=year_count.index,
    y=year_count.values,
    title='Content Released by Year'
)

fig.update_layout(
    xaxis_title='Release Year',
    yaxis_title='Number of Titles'
)

fig.show()
```

**IMDb Rating Distribution**

```python
fig = px.histogram(
    titles,
    x='imdb_score',
    title='IMDb Rating Distribution'
)

fig.show()
```

**Average IMDb Rating — Movies vs Shows**

```python
rating = titles.groupby('type')['imdb_score'].mean().reset_index()

fig = px.bar(
    rating,
    x='type',
    y='imdb_score',
    title='Average IMDb Rating by Content Type'
)

fig.show()
```

**Top 10 Highest-Rated Titles**

```python
fig = px.bar(
    top_rated,
    x='imdb_score',
    y='title',
    orientation='h',
    title='Top 10 Highest Rated Titles'
)

fig.show()
```

**Top 10 Most Popular Titles**

```python
fig = px.bar(
    popular,
    x='tmdb_popularity',
    y='title',
    orientation='h',
    title='Top 10 Most Popular Titles'
)

fig.show()
```

**Movie Runtime Distribution**

```python
fig = px.histogram(
    movies,
    x='runtime',
    title='Movie Runtime Distribution'
)

fig.update_layout(
    xaxis_title='Runtime in Minutes',
    yaxis_title='Number of Movies'
)

fig.show()
```

**Age Certification**

```python
age = titles['age_certification'].value_counts().head(10)

fig = px.bar(
    x=age.index,
    y=age.values,
    title='Top Age Certifications'
)

fig.update_layout(
    xaxis_title='Age Certification',
    yaxis_title='Number of Titles'
)

fig.show()
```

**Top 10 Actors**

```python
top_actors = actors['name'].value_counts().head(10)

fig = px.bar(
    x=top_actors.values,
    y=top_actors.index,
    orientation='h',
    title='Top 10 Actors'
)

fig.update_layout(
    xaxis_title='Number of Appearances',
    yaxis_title='Actor'
)

fig.show()
```

**Top 10 Directors**

```python
top_directors = directors['name'].value_counts().head(10)

fig = px.bar(
    x=top_directors.values,
    y=top_directors.index,
    orientation='h',
    title='Top 10 Directors'
)

fig.update_layout(
    xaxis_title='Number of Titles',
    yaxis_title='Director'
)

fig.show()
```
